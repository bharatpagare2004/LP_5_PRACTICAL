```python
"""1.  Linear regression by using Deep Neural network: Implement Boston housing price prediction
problem by Linear regression using Deep Neural network. Use Boston House price prediction
dataset."""

import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

#loading data

df = pd.read_csv("HousingData.csv")
df.head(5)

#fill missing data

df = df.fillna(df.mean())
df.head(5)


#input and target
x = df.drop("MEDV",axis = 1)
y =df["MEDV"]

#spliting data into training and testing

x_train,x_test,y_train ,y_test = train_test_split(x,y,test_size = 0.2,random_state= 42)

#feature scaling
scaler = StandardScaler()
x_train = scaler.fit_transform(x_train)
x_test = scaler.transform(x_test)

#model creation

model = keras.Sequential([

                         keras.layers.Dense(64,activation = 'relu',input_shape=(x_train.shape[1],)),
                         keras.layers.Dense(32,activation = 'relu'),
                         keras.layers.Dense(1)
])

#model compile
model.compile(
    optimizer = 'adam',
    loss= 'mse',
    metrics = ['mae']

)

#model train

history = model.fit(x_train,y_train,
    validation_split=0.2,
    epochs = 100,
    verbose = 1
)

#model evaluate
loss , mae = model.evaluate(x_test,y_test)
print("mae",mae)

df.head(5)


#prediction
print("enter the parameter for prediction")

CRIM = float(input("enter the CRIM: "))
ZN = float(input("enter the ZN: "))
INDUS = float(input("enter the INDUS: "))
CHAS = float(input("enter the CHAS: "))
NOX = float(input("enter the NOX: "))
RM = float(input("enter the RM: "))
AGE = float(input("enter the AGE: "))
DIS = float(input("enter the DIS: "))
RAD = float(input("enter the RAD: "))
TAX = float(input("enter the TAX: "))
PTRATIO = float(input("enter the PTRAITO3: "))
B = float(input("enter the B: "))
LSTAT = float(input("enter the LSTAT: "))

user_data = np.array([[CRIM,	ZN,	INDUS,	CHAS	,NOX,	RM,	AGE,	DIS,	RAD,	TAX,	PTRATIO,	B	,LSTAT]])

user_data = scaler.transform(user_data)

prediction = model.predict(user_data)
print("the predicted value is:",prediction)
```
